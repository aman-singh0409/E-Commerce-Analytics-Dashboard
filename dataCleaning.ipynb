# Loading Dataset

```python
import pandas as pd
```

```python
df = pd.read_csv(
    "data/data.csv",
    encoding='latin1'
)
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df = df.drop_duplicates()
```

```python
df = df[~df["InvoiceNo"].astype(str).str.contains("C")]
```

```python
df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df.to_csv(
 "data/cleaned_data.csv",
 index=False
)
```
