# Revenue Anlaysis

```python
import pandas as pd

df = pd.read_csv("data/data.csv", encoding="latin1")
```

```python
print(df.columns)
```

```python
df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]
```

```python
df["TotalPrice"].sum()
```

# Monthly Revenue

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
monthly_sales = df.groupby(
    df["InvoiceDate"].dt.month
)["TotalPrice"].sum()
```

# Top Customers

```python
top_customers = df.groupby(
 "CustomerID"
)["TotalPrice"].sum().sort_values(
 ascending=False
)
```

# Top products

```python
Top_products = df.groupby(
 "Description"
)["Quantity"].sum().sort_values(
 ascending=False
)
```

# Repeat Buyer Analysis

```python
repeat_buyers = df.groupby(
 "CustomerID"
)["InvoiceNo"].nunique()
repeat_buyers = repeat_buyers[
 repeat_buyers > 1
]
```

## Repeat Buyer %

```python
repeat_percentage = (
 len(repeat_buyers) /
 df["CustomerID"].nunique()
) * 100
```
