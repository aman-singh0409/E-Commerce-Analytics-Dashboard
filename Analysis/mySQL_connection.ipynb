```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
df = pd.read_csv("data/data.csv", encoding="latin1")
```

```python
engine = create_engine(
    "mysql+pymysql://root:sql%40aman0409@localhost/ecommerce"
)
```

```python
df.to_sql(
    name="orders",
    con=engine,
    if_exists="replace",
    index=False
)
```
